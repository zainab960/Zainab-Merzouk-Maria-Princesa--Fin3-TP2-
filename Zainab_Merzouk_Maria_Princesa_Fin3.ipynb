```python
import sqlite3

# Connexion à la base de données SQLite (ou création d'un fichier local)
conn = sqlite3.connect("chomage.db")
cursor = conn.cursor()

# 1. Création de la table
cursor.execute(
    """
CREATE TABLE IF NOT EXISTS duree_chomage_national (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    annee INTEGER NOT NULL UNIQUE,
    chomage_0_11_mois REAL NOT NULL,
    chomage_12_mois_plus REAL NOT NULL,
    source TEXT DEFAULT "Enquête nationale sur l'emploi, Haut Commissariat au Plan"
);
"""
)

# 2. Liste des données à insérer
donnees = [
    (1999, 28.43, 71.47),
    (2000, 29.89, 70.03),
    (2001, 30.91, 69.08),
    (2002, 30.49, 69.50),
    (2003, 29.37, 70.62),
    (2004, 30.29, 69.64),
    (2005, 31.42, 68.27),
    (2006, 34.92, 65.05),
    (2007, 33.92, 65.79),
    (2008, 30.54, 69.43),
    (2009, 33.73, 65.96),
    (2010, 36.87, 62.81),
    (2011, 34.97, 64.80),
    (2012, 34.53, 65.09),
    (2013, 35.83, 63.97),
    (2014, 37.95, 61.76),
    (2015, 34.50, 65.30),
    (2016, 32.60, 67.20),
    (2017, 28.60, 71.20),
    (2018, 32.10, 67.90),
    (2019, 31.80, 68.20),
    (2020, 43.70, 56.30),
    (2021, 27.50, 72.50),
    (2022, 31.50, 68.50),
    (2023, 33.60, 66.40),
    (2024, 37.60, 62.10),
    (2025, 35.20, 64.80),
]

# 3. Insertion des données en bloc
cursor.executemany(
    """
    INSERT OR IGNORE INTO duree_chomage_national (annee, chomage_0_11_mois, chomage_12_mois_plus)
    VALUES (?, ?, ?);
    """,
    donnees,
)

# Validation et fermeture
conn.commit()
conn.close()

print("Table créée et données insérées avec succès dans SQLite !")
```

```python
# Section 5: Création de la Base de Données SQLite
import sqlite3

def creer_base_de_donnees():
    # Connexion à la base de données SQLite
    conn = sqlite3.connect("chomage.db")
    cursor = conn.cursor()

    # Création de la table pour stocker la durée du chômage au niveau national
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS duree_chomage_national (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        annee INTEGER NOT NULL UNIQUE,
        chomage_0_11_mois REAL NOT NULL,
        chomage_12_mois_plus REAL NOT NULL,
        source TEXT DEFAULT "Enquête nationale sur l'emploi, Haut Commissariat au Plan"
    );
    """)

    conn.commit()
    conn.close()
    print("Base de données SQLite initialisée et table créée avec succès !")

creer_base_de_donnees()
```

```python
# Section 6: Description de la Structure de la Base de Données
import sqlite3
import pandas as pd

def decrire_structure_base():
    conn = sqlite3.connect("chomage.db")
    cursor = conn.cursor()

    # Récupération des informations sur les colonnes de la table
    cursor.execute("PRAGMA table_info(duree_chomage_national);")
    colonnes = cursor.fetchall()

    # Conversion en DataFrame pour un affichage propre
    df_structure = pd.DataFrame(colonnes, columns=["Index", "Nom de Colonne", "Type", "Non Null", "Valeur par Défaut", "Clé Primaire"])

    print("--- Structure de la table 'duree_chomage_national' ---")
    display(df_structure)

    conn.close()

decrire_structure_base()
```

### Question 9 : Moyenne annuelle de l'indice de la 'Santé' pour chaque année

### Question 10 : Différence d'indice pour 'Produits alimentaires' entre décembre 2025 et janvier 2026
